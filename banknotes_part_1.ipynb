## Problem 1

Using  the [GaussianNB](https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.GaussianNB.html) function construct the  Gaussian  Bayes classifier using only one feature. Which feature will you choose? Calculate the confusion matrix (normalized as to show rates), ROC AUC score and plot ROC curve. Do this both for training and validation set. Plot both curves on the same plot.  

```python
from sklearn.naive_bayes import GaussianNB
```

__Hint__ For calculating metrics and plotting ROC curves you may use functions from scikit-learn: `roc_curve`, `roc_auc_score` and `confusion matrix`. For estimating normal distribution parameters  use `norm.fit` `from scipy.stats`. Use `norm.pdf` for normal probability density function.

```python
from sklearn.metrics import roc_auc_score, roc_curve, confusion_matrix
```

Which feature did you choose?

```python
# a0 - it gives the best one-feature ROC AUC on the set
feature_scores = []
for feature in data.columns[:-1]:
    model = GaussianNB()
    model.fit(data_train[[feature]], lbls_train)
    train_proba = model.predict_proba(data_train[[feature]])[:, 1]
    test_proba = model.predict_proba(data_test[[feature]])[:, 1]
    feature_scores.append({
        "feature": feature,
        "train_auc": roc_auc_score(lbls_train, train_proba),
        "validation_auc": roc_auc_score(lbls_test, test_proba),
    })

feature_scores = pd.DataFrame(feature_scores).sort_values("train_auc", ascending=False)
feature_scores
```

```python
selected_feature = "a0"
gnb = GaussianNB()
gnb.fit(data_train[[selected_feature]], lbls_train)

train_pred = gnb.predict(data_train[[selected_feature]])
test_pred = gnb.predict(data_test[[selected_feature]])
train_proba = gnb.predict_proba(data_train[[selected_feature]])[:, 1]
test_proba = gnb.predict_proba(data_test[[selected_feature]])[:, 1]

train_cm = confusion_matrix(lbls_train, train_pred, normalize="true")
test_cm = confusion_matrix(lbls_test, test_pred, normalize="true")
train_auc = roc_auc_score(lbls_train, train_proba)
test_auc = roc_auc_score(lbls_test, test_proba)
fprs_train, tprs_train, _ = roc_curve(lbls_train, train_proba)
fprs_test, tprs_test, _ = roc_curve(lbls_test, test_proba)

print(f"Chosen feature: {selected_feature}")
print(f"Train AUC: {train_auc:.4f}")
print(f"Validation AUC: {test_auc:.4f}")
print("Train confusion matrix, normalized by true class:")
print(np.round(train_cm, 4))
print("Validation confusion matrix, normalized by true class:")
print(np.round(test_cm, 4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
ConfusionMatrixDisplay(train_cm, display_labels=gnb.classes_).plot(
    ax=axes[0], values_format=".3f", colorbar=False
)
axes[0].set_title("Train normalized confusion matrix")
ConfusionMatrixDisplay(test_cm, display_labels=gnb.classes_).plot(
    ax=axes[1], values_format=".3f", colorbar=False
)
axes[1].set_title("Validation normalized confusion matrix")
fig.tight_layout();
```

```python
fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(fprs_train, tprs_train, label=f"train AUC = {train_auc:.4f}")
ax.plot(fprs_test, tprs_test, label=f"validation AUC = {test_auc:.4f}")
ax.plot([0, 1], [0, 1], linestyle="--", color="black", alpha=0.5)
ax.set_xlabel("False positive rate")
ax.set_ylabel("True positive rate")
ax.set_title("GaussianNB ROC curve using a0")
ax.set_aspect(1.0)
ax.legend();
```
